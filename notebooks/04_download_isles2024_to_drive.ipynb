# ISLES'24 persistent download to Google Drive

This notebook stores the current public ISLES'24 training archive directly in Google Drive. The archive is about **99 GB**.

If Colab disconnects, reopen this same notebook and choose **Runtime → Run all**. The downloader resumes the partial file and verifies the published Zenodo v7 MD5 when complete.

Destination:
`MyDrive/HybridStrokeSeg/ISLES2024/raw/train.7z`

This notebook does **not** extract the whole archive yet. The 4-D CTP data dominate the archive size; after the archive is safe in Drive, the project will selectively extract the channels required for each predeclared experiment.


## 1. Mount Google Drive


```python
from google.colab import drive
drive.mount("/content/drive")
```

## 2. Clone the latest repository


```python
import os, shutil, subprocess
from pathlib import Path
REPO = Path("/content/HybridStrokeSeg-Reproduction")
if REPO.exists():
    shutil.rmtree(REPO)
subprocess.run(["git","clone","-q","https://github.com/NTinkicht/HybridStrokeSeg-Reproduction.git",str(REPO)],check=True)
os.chdir(REPO)
subprocess.run(["python","-m","pip","install","-q","-e","."],check=True)
subprocess.run(["git","rev-parse","--short","HEAD"],check=True)
```

## 3. Download or resume the 99 GB ISLES'24 archive


```python
from pathlib import Path
ARCHIVE = Path("/content/drive/MyDrive/HybridStrokeSeg/ISLES2024/raw/train.7z")
ARCHIVE.parent.mkdir(parents=True, exist_ok=True)
subprocess.run(["python","scripts/download_isles2024.py","--archive",str(ARCHIVE)],check=True)
```

## 4. Confirm persistent archive


```python
print("Archive:", ARCHIVE)
print("Exists:", ARCHIVE.exists())
if ARCHIVE.exists():
    print("Size GiB:", round(ARCHIVE.stat().st_size/(1024**3), 2))
print("The downloader only exits successfully after MD5 verification unless explicitly told otherwise.")
```

## Reconnection rule

If the runtime disconnects during the download, just run this notebook again from the top. Do not delete `train.7z`; the partial archive is the resume point.
